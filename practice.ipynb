Single layer

```python
import numpy as np

inputs = [0.5, 0.2, 0.8]
weights = [0.1, 0.3, -0.2]
bias = 0.1

output = np.dot(inputs, weights) + bias
print(output)
```

Multilayer

```python
inputs = [0.5, 0.2, 0.8, 0.1]
weights = [[0.1, 0.2, -0.1],
           [-0.3, 0.4, 0.2],
           [0.5, -0.5, 0.6],
           [-0.2, 0.1, -0.4]]
bias  = [0.1,0.2,0.3]

output = np.dot(inputs, weights) + bias
print(output)
```

Data batch

```python
input_databatch =[
    [0.5, 0.2, 0.8, 0.1],           #1st row is first set of inputs x1,x2,x3,x4
    [0.1, 0.3, -0.2, 0.9],
    [0.9, 0.7, 0.6, 0.4],
    [0.3, 0.8, 0.1, 0.5],
    [0.6, 0.9, 0.3, 0.7],
    [0.2, 0.4, 0.7, 0.2]
]

weight = [
    [0.1, 0.2, -0.1],
    [-0.3, 0.4, 0.2],
    [0.5, -0.5, 0.6],
    [-0.2, 0.1, -0.4]]

bias = [0.1, 0.2, 0.3]

output = np.dot(input_databatch, np.array(weight)) + bias
print(output)

```
